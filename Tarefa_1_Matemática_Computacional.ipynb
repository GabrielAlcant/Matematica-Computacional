```
import numpy as np
import matplotlib.pyplot as plt

# Configuração para plots ficarem mais nítidos e com tamanho padrão
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
plt.rcParams['figure.figsize'] = (6, 6)

def plot_vectors(vectors, colors, labels, title="Transformação de Vetores"):
    """
    Função auxiliar para plotar vetores 2D com origem em (0,0).
    """
    plt.figure(figsize=(7, 7))
    ax = plt.gca()

    # Desenha os eixos
    ax.axhline(0, c='black', lw=0.5)
    ax.axvline(0, c='black', lw=0.5)
    ax.grid(True, linestyle='--')

    origin = np.array([0, 0])

    for vec, color, label in zip(vectors, colors, labels):
        ax.quiver(origin[0], origin[1], vec[0], vec[1],
                  angles='xy', scale_units='xy', scale=1,
                  color=color, label=f"{label} {vec}")

    # Define os limites do gráfico
    all_vecs = np.array(vectors)
    max_val = np.max(np.abs(all_vecs)) * 1.3
    if max_val == 0: max_val = 5 # Caso de fallback

    ax.set_xlim([-max_val, max_val])
    ax.set_ylim([-max_val, max_val])
    ax.set_aspect('equal')
    ax.legend()
    ax.set_title(title)
    plt.show()
```

```
# Vetores da base canônica
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Matriz de Rotação Horária 90°
R = np.array([
    [0, 1],
    [-1, 0]
])

# Vetores da base transformados
T_e1 = R @ e1
T_e2 = R @ e2

print(f"Base original: e1 = {e1}, e2 = {e2}")
print(f"Base transformada: T(e1) = {T_e1}, T(e2) = {T_e2}")

# Plotando
plot_vectors(
    [e1, e2, T_e1, T_e2],
    ['blue', 'blue', 'red', 'red'],
    ['e1 (original)', 'e2 (original)', 'T(e1) (rotac.)', 'T(e2) (rotac.)'],
    "Transformação da Base (Rotação Horária 90°)"
)
```

```
# Vetores arbitrários
v1 = np.array([2, 3])
v2 = np.array([-1, 2])

# Aplicando a rotação R
w1 = R @ v1
w2 = R @ v2

print(f"Vetor v1 = {v1}  --->  w1 = {w1}")
print(f"Vetor v2 = {v2}  --->  w2 = {w2}")

# Plotando
plot_vectors(
    [v1, v2, w1, w2],
    ['blue', 'blue', 'red', 'red'],
    ['v1 (original)', 'v2 (original)', 'w1 (rotac.)', 'w2 (rotac.)'],
    "Aplicação da Rotação a Vetores Arbitrários"
)
```

```
# Vetores da base canônica (já definidos, mas repetindo para clareza)
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Matriz de Cisalhamento em x (k=1.5)
S = np.array([
    [1, 1.5],
    [0, 1]
])

# Vetores da base transformados
T_e1_shear = S @ e1
T_e2_shear = S @ e2

print(f"Base original: e1 = {e1}, e2 = {e2}")
print(f"Base transformada: T(e1) = {T_e1_shear}, T(e2) = {T_e2_shear}")

# Plotando
plot_vectors(
    [e1, e2, T_e1_shear, T_e2_shear],
    ['blue', 'blue', 'red', 'red'],
    ['e1 (original)', 'e2 (original)', 'T(e1) (shear)', 'T(e2) (shear)'],
    "Transformação da Base (Cisalhamento em x, k=1.5)"
)
```
