# PARTE A

Considere a função:

$$f(x) = \cos(x) - x$$


```python
import math

def f(x):
    return math.cos(x) - x

def df(x):
    return -math.sin(x) - 1

def g(x):
    return math.cos(x)

tol = 1e-10
max_iter = 100
```

## Método de Newton–Raphson

```python
# Algoritmo de Newton-Raphson
# 1. Escolha uma aproximação inicial x0
# 2. Para k = 1, 2, 3, ...
# 3.    Calcule f(x_{k-1}) e f'(x_{k-1})
# 4.    x_k = x_{k-1} - f(x_{k-1}) / f'(x_{k-1})
# 5.    Se |x_k - x_{k-1}| < tolerância, pare
# 6. Fim

import time
import pandas as pd

def newton(x0):
    rows = []
    x_prev = x0
    for k in range(1, max_iter+1):
        t0 = time.perf_counter()
        xk = x_prev - f(x_prev)/df(x_prev)
        t1 = time.perf_counter()
        erro = abs(xk - x_prev)
        rows.append([k, xk, f(xk), erro, t1-t0])
        if erro < tol:
            break
        x_prev = xk
    return pd.DataFrame(rows, columns=['k','x_k','f(x_k)','erro','tempo'])

df_newton = newton(0.5)
df_newton
```

O método de Newton apresenta convergência quadrática quando iniciado próximo à raiz, resultando em poucas iterações.

# PARTE B — Adaptação e comparação dos métodos

## Método da Secante

```python
# Algoritmo da Secante
# 1. Escolha duas aproximações iniciais x0 e x1
# 2. Para k = 2, 3, ...
# 3.    x_k = x_{k-1} - f(x_{k-1}) * (x_{k-1}-x_{k-2}) / (f(x_{k-1}) - f(x_{k-2}))
# 4.    Se |x_k - x_{k-1}| < tolerância, pare
# 5. Fim

def secante(x0, x1):
    rows = []
    for k in range(2, max_iter+1):
        t0 = time.perf_counter()
        xk = x1 - f(x1)*(x1-x0)/(f(x1)-f(x0))
        t1 = time.perf_counter()
        erro = abs(xk - x1)
        rows.append([k, xk, f(xk), erro, t1-t0])
        if erro < tol:
            break
        x0, x1 = x1, xk
    return pd.DataFrame(rows, columns=['k','x_k','f(x_k)','erro','tempo'])

df_secante = secante(0.5, 0.7)
df_secante
```

O método da Secante elimina o cálculo da derivada, mantendo convergência superlinear.

## Método do Ponto Fixo (MPF)

```python
# Algoritmo do Ponto Fixo
# 1. Escolha uma aproximação inicial x0
# 2. Para k = 1, 2, 3, ...
# 3.    x_k = g(x_{k-1})
# 4.    Se |x_k - x_{k-1}| < tolerância, pare
# 5. Fim

def ponto_fixo(x0):
    rows = []
    x_prev = x0
    for k in range(1, max_iter+1):
        t0 = time.perf_counter()
        xk = g(x_prev)
        t1 = time.perf_counter()
        erro = abs(xk - x_prev)
        rows.append([k, xk, erro, t1-t0])
        if erro < tol:
            break
        x_prev = xk
    return pd.DataFrame(rows, columns=['k','x_k','erro','tempo'])

df_ponto_fixo = ponto_fixo(0.5)
df_ponto_fixo
```

## Comparação dos Métodos

| Método | Iterações | Avaliações de Função | Convergência |
|--------|-----------|---------------------|--------------|
| Newton | Baixa | f e f' | Quadrática |
| Secante | Média | Apenas f | Superlinear |
| MPF | Alta | Apenas g | Linear |
